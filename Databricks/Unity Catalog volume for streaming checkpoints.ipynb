This sequence creates and verifies your Unity Catalog volume for streaming checkpoints.

You can reuse the path in your streaming script as:

python

The display() calls confirm each step visually in the notebook output.

```python
# 1️⃣ Show existing schemas
display(spark.sql(""" SHOW SCHEMAS IN dbw_weather_straming """))

# 2️⃣ Create the volume if it doesn’t exist
spark.sql("""
CREATE VOLUME IF NOT EXISTS
dbw_weather_straming.default.weather_volume
COMMENT 'Volume for weather streaming checkpoints'
""")

# 3️⃣ Verify the volume creation
display(spark.sql(""" SHOW VOLUMES IN dbw_weather_straming.default """))

# 4️⃣ Define the volume path and list contents
volume_path = "/Volumes/dbw_weather_straming/default/weather_volume"
display(dbutils.fs.ls(volume_path))

# 5️⃣ Create the checkpoint directory inside the volume
dbutils.fs.mkdirs(f"{volume_path}/checkpoints/weather_stream")
```
